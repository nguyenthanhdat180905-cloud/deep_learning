Bài 8 : autograd cơ bản 
 - yêu cầu : 
   - tính thủ công đạo hàm  y = x**2 tại x = 3.6 , so sánh x.grad 
   - đổi sang hàm y = 3*x**2 +2*x+1 , kiểm tra gradient 

```python
x = torch.tensor(
    3.6,
    requires_grad=True
)
y = x * x
y.backward()
print("x =", x)
print("y =", y)
print("x.grad =", x.grad)
```

  - requires_grad=True không phải là đạo hàm mà là PyTorch theo dõi các phép tính liên quan đến x để sau này có thể tính gradient.

Bài 9 : gradient descent tìm hệ số đa thức 
 - yêu cầu :
   - so sánh ba hệ số học được với [1,2,3]
   - giải thích vai trò optimizer.zero_grad(), mse.backward() và optimizer.step().

```python
import numpy as np
import torch
polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
    ])
w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)
print("Initial coefficients:")
print(w)
for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()
print("Learned coefficients:")
print(w)
```

Answer: 
| Hệ số  | Đúng | Học được | Sai lệch |
| ------ | ---: | -------: | -------: |
| `w[0]` |    1 |   1.0270 |  +0.0270 |
| `w[1]` |    2 |   1.9322 |  -0.0678 |
| `w[2]` |    3 |   1.8358 |  -1.1642 |

- optimizer.zero_grad xóa gradient của lần lặp trước.
- mse.backward() lan truyền ngược (backpropagation) để tính đạo hàm của mse theo các hệ số w. 
- optimizer.step() gradient vừa tính để cập nhật các hệ số w.

Bài 10  : giải hệ phương trình bằng autograd 
 - yêu cầu : 
   - kiểm tra nghiệm thu được với A+B=9 , C-D=1 , A+C=8 , B-D=2
   - nhận xét vì sao bài toán có thể có nhiều nghiệm 
   

```python
import random
import torch
A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)
EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)
for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()
print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)
```

Answer:
$$A+B=9$$

suy ra:

$$ B=9-A $$

Từ (3):

$$ A+C=8 $$

suy ra:

$$ C=8-A $$

Từ (2):

$$ C-D=1 $$

suy ra:

$$ D=C-1=7-A $$

Kiểm tra phương trình (4):

B−D=(9−A)−(7−A)=2

Bài toán không có một nghiệm duy nhất mà có vô số nghiệm, vì 4 phương trình không độc lập hoàn toàn. Phương trình thứ tư có thể được suy ra từ ba phương trình còn lại.
